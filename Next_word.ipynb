```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
token=Tokenizer()
```

```python
token.fit_on_texts([data])
```

```python
token.word_index
```

```python
len(token.word_index)
```

```python
input_sequences = []
for sentence in data.split('\n'):
  tokenized_sentence = token.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

```python
max_leng=max([len(x) for x in input_sequences])
```

```python
max_leng
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
padded_input_sequences=pad_sequences(input_sequences, maxlen=max_leng,padding='pre')
```

```python
X=padded_input_sequences[:,:-1]
```

```python
X
```

```python
y=padded_input_sequences[:,-1]
```

```python
y
```

```python
X.shape
```

```python
y.shape
```

```python
from tensorflow.keras.utils import to_categorical
y=to_categorical(y,num_classes=524)
```

```python
y
```

```python
from tensorflow.keras.layers import Input, Embedding, LSTM, Dense
from tensorflow.keras.models import Model

maxlen = 31
max_features = 524
embed_size = 100

inp = Input(shape=(maxlen, ))

x = Embedding(input_dim=max_features, output_dim=embed_size)(inp)

x = LSTM(150)(x)

x = Dense(524, activation="softmax")(x)

model = Model(inputs=inp, outputs=x)

model.compile(loss='categorical_crossentropy',
              optimizer='adam',
              metrics=['accuracy'])

model.summary()
```

```python
model.fit(X,y,epochs=10)
```
